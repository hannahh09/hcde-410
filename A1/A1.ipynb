## Part 0 — Get the data

The data comes from [data.seattle.gov](https://data.seattle.gov/), the City of Seattle's open-data portal. We'll use the [Burke-Gilman Trail counter north of NE 70th St](https://data.seattle.gov/Transportation/Burke-Gilman-Trail-north-of-NE-70th-St-Bicycle-and/2z5v-ecg8), which records **bike and pedestrian traffic, by hour, by direction**.

We'll pull the data straight from the city's API rather than downloading a file by hand. That's a small reproducibility win: anyone who runs this notebook gets the same data the same way, and it's written down exactly where it came from.

````python
import requests
import pandas as pd
import matplotlib.pyplot as plt
````

````python
# The dataset's API endpoint, and a query for one year of data (2019).
# You can paste the full URL into a browser to preview the raw data.
api_endpoint = "https://data.seattle.gov/resource/2z5v-ecg8.json?"
api_parameters = "$limit=50000&$where=date > '2019-01-01T00:00:00' AND date < '2020-01-01T00:00:00'"

response = requests.get(api_endpoint + api_parameters)
raw_data = response.json()
print("Downloaded", len(raw_data), "hourly records.")
# If this cell fails (e.g. the API is down), see the note below the next cell.
````

Now load it into a **pandas DataFrame** — a table, like a spreadsheet you can compute with.

````python
df = pd.DataFrame(raw_data)
df.head()
````

### Quick sanity checks

Whenever you work with data **you didn't create**, check that it means what you think before you analyze it. Two things need fixing here:
- the counts arrive as **text**, so we convert them to numbers;
- the `date` is a string, so we convert it to a real **datetime** and pull out the hour and weekday.

We'll also add tidy `bikes` and `peds` columns (north + south).

````python
# convert the count columns from text to numbers
count_cols = ["bike_north", "bike_south", "ped_north", "ped_south", "bgt_north_of_ne_70th_total"]
df[count_cols] = df[count_cols].apply(pd.to_numeric, errors="coerce")

# convert the date string to a real datetime, and derive hour + weekday
df["date"] = pd.to_datetime(df["date"])
df["hour"] = df["date"].dt.hour
df["weekday"] = df["date"].dt.day_name()

# combined bike and pedestrian counts
df["bikes"] = df["bike_north"] + df["bike_south"]
df["peds"]  = df["ped_north"] + df["ped_south"]

df.info()
df.describe()
````

**Look at what `describe()` tells you.** Are there hours with 0 traffic? Any missing values (a sensor can go offline)? Note anything odd — you'll want to mention data limitations later.

## Part 1 — One worked example: when do cyclists ride?

Here is the whole technique, **once**, so you can see the pattern. After this, the questions are yours.

The move is always the same: **group → summarize → plot → interpret.** In pandas that's a couple of lines (no loops needed).

````python
# group every row by its hour of day, and sum bikes and peds within each hour
by_hour = df.groupby("hour")[["bikes", "peds"]].sum()

by_hour.plot(figsize=(11, 5))
plt.title("Burke-Gilman traffic by hour of day (2019)")
plt.xlabel("Hour of day (0–23)")
plt.ylabel("Total count in 2019")
plt.xticks(range(0, 24))
plt.show()
````

**Now read the plot — this is the actual work.** What does the shape tell you? Do you see one peak or two? What might a double peak mean about *why* people are on the trail? That interpretation — turning a shape into a claim about human behavior — is what every question below is really asking for.

**Q1. When is the trail busiest, and when is it quietest?**
Decide what *busy* means to you first — a single hour? a stretch of the day? bikes and peds together, or separately? — and say why you chose that.

````python
by_hour = df.groupby("hour")[["bikes","peds"]].mean()
by_hour["total"] = by_hour["bikes"] + by_hour["peds"]
print(by_hour["total"].idxmax(), by_hour["total"].idxmin())
print(by_hour.round(1).loc[[2, 8, 9, 17]])
by_hour.plot(figsize=(11,5)); plt.xticks(range(24))
plt.title("Average bikes and pedestrians per hour of day (2019)")
plt.xlabel("Hour of day (0–23)"); plt.ylabel("Average people per hour")
plt.show()
````

**Your Q1 answer:**

I define *busy* as the hour with the highest average count of bikes and pedestrians combined, because the counter records hourly and both groups share the use of the trail. Averaged across every day of 2019, the busiest hour is 5 pm (102 bikes, and 27 pedestrians), while the quietest hour is 2 am (close to 0 bikes and pedestrians), though every hour from midnight to 3 am have counts close to 0. Reading the graph, traffic reaches about 100 people per hour by 8 am and stays near that level throughout midday, so while 5 pm is the peak, there is steady daytime use of the trail. Pedestrians peak at 9 am (39), so the 5 pm busy peak is driven by bikes, which make up most of the combined count. 

**Q2. Do cyclists and pedestrians use the trail at different times of day?**
Compare their hourly patterns. If they differ, what might explain it?

````python
by_hour[["bikes","peds"]].plot(figsize=(11,5)); plt.xticks(range(24))
plt.title("Average bikes and pedestrians per hour of day (2019)")
plt.xlabel("Hour of day (0–23)"); plt.ylabel("Average people per hour")
plt.show()
print(by_hour.round(0).loc[[8, 9, 17], ["bikes","peds"]])
````

**Your Q2 answer:**

Yes, cyclists and pedestrians use the trail at different times of day. Bikes have two peaks, one at 8 am (67) and a larger one at 5 pm (102), with a dip at 9 am (59). Pedestrians peak once, at 9 am (39), and stay fairly steady through midday, with 27 at 5 pm. The hours suggest cyclists ride at the start and end of the workday, which fits commuting, while pedestrians use the trail at more flexible times, such as for walks or exercise. The counter doesn’t record why people are on the trail, so these are possible explanations from my own speculations.

**Q3. Commuting or recreation?**
Is there evidence people use the Burke-Gilman to *commute*, not just for fun? Compare **weekdays vs. weekends** (we made a `weekday` column for you), and maybe morning vs. evening. Define what a "commuting pattern" would look like in the data, then say whether you see it.

*Hint: `df.groupby("weekday")[["bikes","peds"]].sum()` gets you started. Weekends and weekdays will look different if commuting is real.*

````python
df["is_weekend"] = df["weekday"].isin(["Saturday","Sunday"])
by_hour_type = df.groupby(["is_weekend","hour"])[["bikes","peds"]].mean().unstack(0)
by_hour_type.columns = ["Bikes weekday","Bikes weekend","Peds weekday","Peds weekend"]

by_hour_type.plot(figsize=(11,5)); plt.xticks(range(24))
plt.title("Average bikes and pedestrians per hour: weekdays vs. weekends (2019)")
plt.xlabel("Hour of day (0–23)"); plt.ylabel("Average people per hour")
plt.show()
print(by_hour_type.round(1).loc[[8, 9, 10, 13, 17]])
````

**Your Q3 answer:**

Yes, there is evidence of commuting on weekdays for bikes but not pedestrians from the graph. I defined a commuting pattern in the data as two weekday peaks during morning and evening hours, with a dip at midday, and no matching pattern on weekends. Weekday bike traffic peaks at 8 am (~72) and 5 pm (111) and dips at 10 am (43), while weekend bike traffic has one broad peak around 1 pm (~146). Weekday pedestrian counts stay fairly flat, with only small bumps near 9 am and 4 pm on the graph, while weekend pedestrians are higher in the morning (~64 at 9 am). This suggests cyclists use the trail to commute on weekdays, while pedestrians use it more for recreation. The weekend bike peak is also higher than either weekday peak, so it is also used for recreation. 

**Q4. Direction and flow.**
The counter records **north vs. south** separately. Are mornings mostly one direction and evenings the other? What would that tell you about who's using the trail and where they're heading?

````python
wk = df[~df["is_weekend"]].groupby("hour")[["bike_north","bike_south","ped_north","ped_south"]].mean()

wk.plot(figsize=(11,5)); plt.xticks(range(24))
plt.title("Average weekday bikes and pedestrians per hour by direction (2019)")
plt.xlabel("Hour of day (0–23)"); plt.ylabel("Average people per hour")
plt.show()
print(wk.round(1).loc[[8, 17]])
````

**Your Q4 answer:**

Yes for bikes, mornings are mostly north and evenings south, but pedestrians show no clear pattern. On weekdays at 8 am about 54 bikes go north and 18 go south, while at 5 pm about 81 go south and 31 go north. Pedestrians stay close to even in both directions, with 14 north and 12 south at 8 am, and 13 north and 14 south at 5 pm. The bike pattern, on average across weekdays, suggests a commuting pattern, since leisure bikers are less likely to all turn around at the same hours. It suggests many weekday cyclists head north in the morning and south in the evening on round trips. However, a limitation is that the counter only sees one point on the trail, so it can’t show where people start or end their trips.
