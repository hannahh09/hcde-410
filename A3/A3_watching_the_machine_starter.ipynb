*Bring the probe and the observation together into one account. What does this mean for trustworthy, reproducible research — and what will you change about how you use and verify AI in your own work?*

For this probe, I gave the same prompt, asking for three peer-reviewed papers on HCI and social media design with DOIs, to Gemini Pro (three runs), ChatGPT (one run), and a participant, P1, using Gemini Flash (one run). The five runs returned 13 distinct papers in 15 slots, with repeats only between my first two Gemini Pro runs, so the same prompt gives a different slice of the literature each time and is not reproducible. I veified each run by opening every DOI, and comparing titles, authors, and year, which only confirms the papers exist. Furthermore, every output looked equally confident, but run 3 (Gemini Pro) included a healthcare paper outside the social media design topic, and P1’s three papers share two authors and one venue (CHI). 

My coded notes show P1 have a high trust towards their Gemini model, as they only verified the first source's DOI and did not open the second or third. I coded it as Partial verification, and they also said they "do not usually ask for DOIs or search outside the chat." This suggests that they would have prompted the AI differently if given the same goal, so I coded it as a prompt mismatch and that their verification process also differed from mine. After the observation, I noticed that all three papers came from CHI and share two authors, suggesting there is a source and sample skew. However, the participant did not mention or notice it during the task.

For trustworthy, reproducible research, this means a DOI gives provenance for a paper but not for why the model selected it, so trust built on format and one check is not enough to examine how the list can potentially be skewed or narrow whose work I cite. Before the task, I wrote about my prior assumptins that AI search is comprehensive because they have a lot of data from the internet, but this activity showed that human oversight matters in that search process. I will change how I use AI for desk research by writing my criteria (peer-reviewed, on topic, varied authors and venues) before prompting, and also ask the AI to explain why it chose each source.
